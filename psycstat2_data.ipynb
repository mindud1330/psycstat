```python
import pandas as pd
import statsmodels.api as sm
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# CSV file path
file_path = "psycstat2_assn1.csv"

# CSV file
df = pd.read_csv(file_path)
df.head()
```

```python
print(df.isnull().sum())
```

```python
subset_df = df[['Ext', 'Neu', 'Agr', 'SS', 'LPM', 'SSz', 'LPMz', 'Alc3']]
```

```python
sns.pairplot(subset_df)
plt.show()
```

```python
correlation_matrix = subset_df.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap='coolwarm', square=True)
plt.title('Correlation Matrix')
plt.show()
```

```python
X = subset_df['LPMz']
Y = subset_df['Alc3']

X = sm.add_constant(X)
LinearModel = sm.OLS(Y, X).fit()

print(LinearModel.summary())
```

```python
X = subset_df[['Agr', 'LPMz']]
Y = subset_df['Alc3']

X_sm = sm.add_constant(X)
LinearModel = sm.OLS(Y, X_sm).fit()

print(LinearModel.summary())
```

```python
X = subset_df[['Ext', 'Agr', 'LPMz']]
Y = subset_df['Alc3']

X_sm = sm.add_constant(X)
LinearModel = sm.OLS(Y, X_sm).fit()

print(LinearModel.summary())
```

```python
X = subset_df[['Ext', 'Neu', 'Agr', 'LPMz']]
Y = subset_df['Alc3']

X_sm = sm.add_constant(X)
LinearModel = sm.OLS(Y, X_sm).fit()

print(LinearModel.summary())
```
