```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf
```

```python
# CSV file path
file_path = "psycstat2_assn2_2024.csv"

# CSV file
df = pd.read_csv(file_path)
df.head()
```

```python
df['d'] = np.where(df['c2'] == "A", 1, np.where(df['c2'] == "B", 0, -99))
df.head()
```

```python
model = smf.ols('y ~ x1 * d', data=df).fit()
print(model.summary())
```
