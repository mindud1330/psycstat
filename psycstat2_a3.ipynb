```python
import pandas as pd
import statsmodels.api as sm
from scipy.stats import f_oneway
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm
```

```python
file_path = "exercise17_19.csv"
data = pd.read_csv(file_path)
print(data.head())
```

```python
data.describe()
```

```python
groups = data['Education'].unique()
scores_by_edu = [data[data['Education'] == group]['Score'] for group in groups]
```

```python
f_stat, p_value = f_oneway(*scores_by_edu)
```

```python
print(f"F-statistic: {f_stat}")
print(f"P-score: {p_value}")
if p_value < 0.05:
    print("그룹 간 평균 차이가 유의합니다.")
else:
    print("그룹 간 평균 차이가 유의하지 않습니다.")
```

```python
groups = data['Group'].unique()
scores_by_group = [data[data['Group'] == group]['Score'] for group in groups]
```

```python
f_stat, p_value = f_oneway(*scores_by_group)
```

```python
print(f"F-statistic: {f_stat}")
print(f"P-value: {p_value}")
if p_value < 0.05:
    print("그룹 간 평균 차이가 유의합니다.")
else:
    print("그룹 간 평균 차이가 유의하지 않습니다.")
```

```python
model = ols('Score ~ C(Education) * C(Group)', data=data).fit()
```

```python
anova_results = anova_lm(model, typ=2)  # Type II ANOVA Table
print(anova_results)
```
