```python
import pandas as pd
import statsmodels.api as sm
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# CSV file path
file_path = "psycstat2_assn1.csv"

# CSV file
df = pd.read_csv(file_path)
df.head()
```

```python
print(df.isnull().sum())
```

```python
# Calculate 
lpm_avg = df['LPM'].mean()
lpm_var = df['LPM'].var()
lpm_sd = df['LPM'].std()

# print result
print(f"average: {lpm_avg}")
print(f"variance: {lpm_var}")
print(f"standard deviation: {lpm_sd}")
```

```python
X1 = df['LPMz']
Y = df['Alc3']

X1 = sm.add_constant(X1)
LinearModel = sm.OLS(Y, X1).fit()

print(LinearModel.summary())
```

```python
X2 = df['LPM']

X2 = sm.add_constant(X2)
LinearModel = sm.OLS(Y, X2).fit()

print(LinearModel.summary())
```

```python
sns.set(style='whitegrid')
sns.regplot(x = 'LPMz', y = 'Alc3', data=df)
plt.xlabel('LPMz')
plt.ylabel('Alc3')

plt.show()
```
